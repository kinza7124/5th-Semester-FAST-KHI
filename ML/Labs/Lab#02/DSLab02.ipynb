**Important Concept: View vs Copy ⚠**

NumPy slicing returns a view, not a copy.

Example:
```
a = np.array([1,2,3,4])
b = a[1:3]

b[0] = 100
print(a)

Output:
[1 100 3 4]
```


WHY?
Because slicing shares memory.

````python
print("Slicing creates a view: ")
arr10= np.array([1,2,3,4,5,6,7])
print(arr10)
b=arr10[1:3]
print(b)

b[0]=100
print(b)
print(arr10)

print("Creating a copy: ")
# create copy
arr10= np.array([1,2,3,4,5,6,7])
print(arr10)
b=arr10[1:3].copy()
print(b)

b[0]=100
print(b)
print(arr10) # original arr10 remained same
````

**What is Reshaping?**

Reshaping means changing the shape (dimensions) of an array without changing the data.

Example:
```
import numpy as np
arr = np.array([1,2,3,4,5,6])

Currently:
Shape: (6,)   → 1D array
Reshape into 2×3 matrix
arr2 = arr.reshape(2,3)
print(arr2)

Output:
[[1 2 3]
 [4 5 6]]

Now shape is:
(2,3)

⚠ Important Rule
Total elements must match.
rows×columns=total elements

Example:
6 elements → can reshape into:

(2,3) ✅
(3,2) ✅
(6,1) ✅
(4,2) ❌ (needs 8 elements)
```
**Automatic Dimension (-1 trick)**
```
arr.reshape(2,-1)
```
NumPy automatically calculates missing dimension.
Very common in ML preprocessing.

````python
arr13 = np.array([1,2,3,4,5,6])
print(arr13.shape)
arr14=arr13.reshape(2,3)
print(arr14)
arr14[0,0]=18
print(arr14)
print(arr13)
# reshape creates a view (not copy).
````

````python
arr15=np.array([[1,2,3],[4,5,6]])
print(arr15)
arr15_flat=arr15.flatten()
print(arr15_flat)
arr15_flat[0]=15
print(arr15_flat)
print(arr15)
# flatten creates a copy that's why original arr15 is not modified
````

````python
arr15=np.array([[1,2,3],[4,5,6]])
print(arr15)
arr15_ravel=arr15.ravel()
print(arr15_ravel)
arr15_ravel[0]=15
print(arr15_ravel)
print(arr15)
# ravel creates a view that's why original arr16 is modified
````

**What is Python GIL (Global Interpreter Lock)? (Advanced)**

Definition:

GIL is a mutex lock in CPython that allows only one thread to execute Python bytecode at a time.

Ensures memory management is safe but limits CPU-bound multithreading.

Implications:

Multithreading in Python is not effective for CPU-bound tasks

Good for I/O-bound tasks (network, file)

For CPU-bound tasks → use multiprocessing

Example:
```
import threading

def func():
    print("Thread running")

t1 = threading.Thread(target=func)
t2 = threading.Thread(target=func)

t1.start()
t2.start()
```
Even though two threads exist, only one executes Python code at a time due to GIL.

GIL exists only in CPython, not in PyPy or Jython.

```
Basic Example
Get even numbers
nums = [1,2,3,4,5,6]
result = filter(lambda x: x % 2 == 0, nums)
print(list(result))

Output:
[2,4,6]
```
Explanation:

Keeps values where condition is True

Removes values where condition is False

**Very Important Concept**

filter():

Does NOT modify original list

Returns an iterator (lazy evaluation)
